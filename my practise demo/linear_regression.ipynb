```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# ---------------------------------------------------------
# 1. Generate Synthetic Data
# ---------------------------------------------------------
np.random.seed(42) # Ensures we get the same random numbers every time

# Generate 50 random ages between 18 and 60
ages = np.random.randint(18, 60, 50)

# Create weights with a linear relationship to age, plus some random noise
# Let's assume base weight is 130 lbs, adding ~1.2 lbs per year of age
weights = 130 + (ages * 1.2) + np.random.normal(0, 10, 50)

# Scikit-learn requires the 'X' (features) to be a 2D array (a column).
# We reshape the 1D ages array into a 2D array.
X = ages.reshape(-1, 1)  
y = weights              

# ---------------------------------------------------------
# 2. Train the Linear Regression Model
# ---------------------------------------------------------
# Initialize the model
model = LinearRegression()

# Fit the model to our data (this calculates the slope and intercept)
model.fit(X, y)

# ---------------------------------------------------------
# 3. Make a Prediction
# ---------------------------------------------------------
# Let's predict the weight of a 35-year-old
test_age = 35

# The predict method also expects a 2D array
predicted_weight = model.predict([[test_age]])

print(f"The equation of the line is: Weight = {model.intercept_:.2f} + {model.coef_[0]:.2f} * Age")
print(f"Predicted weight for a {test_age}-year-old: {predicted_weight[0]:.2f} lbs")

# ---------------------------------------------------------
# 4. Visualize the Results
# ---------------------------------------------------------
# Plot the original data points
plt.scatter(ages, weights, color='blue', alpha=0.6, label='Actual Data')

# Plot the line of best fit (regression line)
# We use the model to predict 'y' values for all our 'X' ages to draw the line
plt.plot(ages, model.predict(X), color='red', linewidth=2, label='Regression Line')

# Highlight our specific prediction
plt.scatter([test_age], predicted_weight, color='green', s=150, zorder=5, label='New Prediction (Age 35)')

plt.title('Predicting Weight based on Age')
plt.xlabel('Age (years)')
plt.ylabel('Weight (lbs)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Our raw dataset
data = pd.DataFrame({
    'Mileage': [25000, 50000, 10000, 80000, 5000],
    'Color': ['Red', 'Blue', 'Red', 'Green', 'Blue'],
    'Price': [22000, 18000, 24000, 12000, 26000] # What we want to predict
})

# 2. Separate our Features (X) from our Target (y)
X = data[['Mileage', 'Color']]
y = data['Price']

# 3. Apply One-Hot Encoding to the 'Color' column
# drop_first=True avoids the Dummy Variable Trap in Linear Regression
X_encoded = pd.get_dummies(X, columns=['Color'], drop_first=True, dtype=int)

print("--- Data Ready for the Algorithm ---")
print(X_encoded)
print("\n")

# 4. Train the Linear Regression Model
model = LinearRegression()
model.fit(X_encoded, y)

# 5. Look at the math it learned!
print("--- The Learned Math ---")
print(f"Base Price (Intercept): ${model.intercept_:.2f}")

# Pair the learned weights (coefficients) with their feature names
for feature, weight in zip(X_encoded.columns, model.coef_):
    print(f"Weight for {feature}: {weight:.2f}")
```

```python
# 1. Import our Machine Learning tools and Pandas
from sklearn.linear_model import LinearRegression
import pandas as pd

# 2. Provide the Historical Data (The "Experience" for our AI)
historical_data = {
    "Open_Price": [100, 150, 200, 250, 300],
    "Close_Price": [105, 152, 208, 255, 310]
}
df = pd.DataFrame(historical_data)

# 3. Separate the Features (What we know) from the Target (What we want to predict)
# scikit-learn requires the Features to be a 2D table, which is why we use double brackets [[ ]]
X = df[["Open_Price"]] # The Feature
y = df["Close_Price"]  # The Target Answer

# 4. Create the "Brain"
# We are choosing a Linear Regression algorithm. It draws the mathematical "line of best fit".
model = LinearRegression()

# 5. Train the AI!
# The .fit() command is the actual learning process. 
# It studies X and y and creates the hidden rules.
model.fit(X, y)

print("--- AI Training Complete! ---")
```

```python
# 6. Ask the AI to predict the close price for a day where the market opens at 350.
# We have to give it the data in the exact same format it was trained on (a DataFrame).
new_day_data = pd.DataFrame({"Open_Price": [350]})

future_prediction = model.predict(new_day_data)

print("If the market opens at 350, the AI predicts it will close at:")
print(future_prediction[0])
```

```python
# 1. Import PyTorch (The standard library for modern AI)
import torch
import torch.nn as nn

print("--- Building a Neural Network ---")

# 2. Create the simplest possible Neural Network
# This is a single layer with 1 input and 1 output.
brain = nn.Linear(in_features=1, out_features=1)

# 3. Create a PyTorch "Tensor" (our data)
# We are going to ask it the exact same question: What happens if the market opens at 350?
input_tensor = torch.tensor([[350.0]])

# 4. Pass the data through the untrained brain
first_guess = brain(input_tensor)

print("The untrained neural network guesses the close price is:")
print(first_guess.item())
```

```python
import torch
import torch.nn as nn

# 1. Prepare the Historical Data as PyTorch Tensors
X_train = torch.tensor([[100.0], [150.0], [200.0], [250.0], [300.0]])
y_train = torch.tensor([[105.0], [152.0], [208.0], [255.0], [310.0]])

# 2. Re-create the Brain (so we start fresh)
# We use a manual "seed" so the random starting weights are exactly the same every time we run it
torch.manual_seed(42) 
brain = nn.Linear(in_features=1, out_features=1)

# 3. Define the Teacher and the Mechanic
# MSELoss = Mean Squared Error (calculates the distance between guess and reality)
loss_function = nn.MSELoss() 
# SGD = Stochastic Gradient Descent (the mechanic that tweaks the weights)
# lr = Learning Rate (how big of an adjustment it makes each time. Too big, and it breaks!)
optimizer = torch.optim.SGD(brain.parameters(), lr=0.00001)

# 4. THE TRAINING LOOP
epochs = 1000 # We will let the AI review the historical data 1,000 times

print("--- Starting AI Training ---")
for epoch in range(epochs):
    # A. Forward Pass: The AI makes its guesses for the historical data
    predictions = brain(X_train)
    
    # B. Calculate Loss: The Teacher grades the guesses
    loss = loss_function(predictions, y_train)
    
    # C. Backpropagation (The Magic): 
    # Clear the old adjustments, calculate the new ones, and update the brain!
    optimizer.zero_grad() 
    loss.backward()       
    optimizer.step()      
    
    # D. Print progress every 200 rounds so we can watch it learn
    if epoch % 200 == 0:
        print("Round " + str(epoch) + " | Error score: " + str(loss.item()))

print("--- AI Training Complete! ---")

# 5. The Final Test
new_day_tensor = torch.tensor([[350.0]])
final_guess = brain(new_day_tensor)

print("\nAfter 1000 rounds of training, the neural network predicts the close price will be:")
print(final_guess.item())
```

```python
import numpy as np
import torch
from sklearn.preprocessing import MinMaxScaler

print("Starting Data Preparation...")

# 1. Extract just the "Close" price column as a NumPy array
# .values.reshape(-1, 1) makes sure the data is in a single vertical column
close_prices = nifty_data['Close'].values.reshape(-1, 1)

# 2. Scale the Data between 0 and 1
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(close_prices)

# 3. Create the Sliding Window (60 days of history to predict 1 day forward)
window_size = 60
X_sequences = [] # This will hold our 60-day chunks (The Features)
y_targets = []   # This will hold the 61st day (The Answer)

# Loop through the data, stopping 60 days before the end
for i in range(len(scaled_data) - window_size):
    # Grab a 60-day chunk
    window = scaled_data[i : i + window_size]
    X_sequences.append(window)
    
    # Grab the very next day
    target = scaled_data[i + window_size]
    y_targets.append(target)

# 4. Convert our Python lists into PyTorch Tensors (so the AI can read them)
# We use torch.float32 because neural networks require 32-bit decimals
X_tensor = torch.tensor(np.array(X_sequences), dtype=torch.float32)
y_tensor = torch.tensor(np.array(y_targets), dtype=torch.float32)

print("--- Data Engineering Complete ---")
print("Total number of 60-day sequences created:")
print(len(X_tensor))
```

```python
import torch.nn as nn

# We define a custom Deep Learning model by creating a Python Class
class TradingLSTM(nn.Module):
    def __init__(self):
        super(TradingLSTM, self).__init__()
        
        # 1. The Memory Layer
        # input_size=1 (We are only looking at Close Price)
        # hidden_size=50 (The AI will track 50 different internal momentum patterns)
        # batch_first=True (Tells PyTorch our data is formatted with the sequence count first)
        self.lstm = nn.LSTM(input_size=1, hidden_size=50, num_layers=1, batch_first=True)
        
        # 2. The Final Prediction Layer
        # It takes the 50 internal momentum signals and outputs 1 final price
        self.linear = nn.Linear(in_features=50, out_features=1)

    # The 'forward' function defines exactly how the data flows through the brain
    def forward(self, x):
        # A. Pass the 60-day window into the LSTM
        lstm_out, _ = self.lstm(x)
        
        # B. Grab the AI's internal thoughts from the very LAST day of the window
        last_day_thoughts = lstm_out[:, -1, :]
        
        # C. Pass those thoughts into the Final Prediction Layer
        final_prediction = self.linear(last_day_thoughts)
        
        return final_prediction

# Initialize the brain!
torch.manual_seed(42)
model = TradingLSTM()

print("--- LSTM Architecture Successfully Initialized ---")
```

```python
print("--- Starting LSTM Training ---")
print("This may take a minute. Watch the error rate drop!")

# 1. The Teacher and The Mechanic
loss_function = nn.MSELoss()
# We use 'Adam' this time. It is a much smarter, modern optimizer than SGD.
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# 2. The Loop (We will train for 50 rounds)
epochs = 50

for epoch in range(epochs):
    # Forward Pass
    predictions = model(X_tensor)
    
    # Calculate Error
    loss = loss_function(predictions, y_tensor)
    
    # Backpropagation (Learn and adjust)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    # Print progress every 10 rounds
    if epoch % 10 == 0:
        print("Epoch " + str(epoch) + " | Error: " + str(loss.item()))

print("--- AI Training Complete! ---")
```

```python
import torch
import matplotlib.pyplot as plt

# 1. Put the model in "Evaluation Mode" (Turns off the training mechanics)
model.eval()

# 2. Predict the past to see how accurate it is
# We use torch.no_grad() to save computer memory since we are no longer training
with torch.no_grad():
    all_predictions_scaled = model(X_tensor).numpy()

# 3. Convert the 0-1 percentages back into real stock prices using our scaler
real_predictions = scaler.inverse_transform(all_predictions_scaled)
real_actuals = scaler.inverse_transform(y_tensor.numpy())

# 4. Predict the FUTURE (Tomorrow's Price)
# Grab the very last 60 days of real market data we have
last_60_days = scaled_data[-60:]
# Reshape it into the exact tensor format the AI expects: [1 batch, 60 days, 1 feature]
future_tensor = torch.tensor(last_60_days.reshape(1, 60, 1), dtype=torch.float32)

with torch.no_grad():
    future_guess_scaled = model(future_tensor).numpy()

# Convert the final guess back to a real price
tomorrow_price = scaler.inverse_transform(future_guess_scaled)

print("--- The AI's Prediction for Tomorrow ---")
print("Nifty 50 Target: " + str(tomorrow_price[0][0]))
print("----------------------------------------\n")

# 5. Visualize the AI's accuracy over the last 100 days
plt.figure(figsize=(10,5))
# Plot the actual market prices in blue
plt.plot(real_actuals[-100:], label="Actual Market Price", color='blue')
# Plot the AI's predictions in red dashed lines
plt.plot(real_predictions[-100:], label="AI Predicted Price", color='red', linestyle='dashed')

plt.title("AI vs Reality: The Last 100 Days")
plt.xlabel("Days (Last 100 Trading Days)")
plt.ylabel("Nifty 50 Price")
plt.legend()
plt.grid(True)
plt.show()
```
